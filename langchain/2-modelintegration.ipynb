### Groq Model Integration

```python
from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": ""
      }
    ],
    temperature=1,
    max_completion_tokens=2048,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

## Sreaming And Batch

### Streaming

#### Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stram() returns an iterator that yiled output chunk as ther are produced. You can use a loop each chunk in real time:

### Summary Comparison

| Setting | What Happens During LLM Streaming |
|---|---|
| `print(chunk.text)` | Words print vertically, one per line. |
| `print(chunk.text, end="")` | Words print horizontally, but might lag/freeze and appear all at once at the end. |
| **`print(chunk.text, end="", flush=True)`** | **True typewriter streaming effect — each token appears on screen smoothly in real-time as it arrives.** |

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("groq:openai/gpt-oss-120b")

model.invoke("Write me a 200 words content on AI")
```

```python
for chunk in model.stream("Why birds are colorful"):
    print(chunk.text, end="", flush=True)
```

### BATCH

#### Batching a collection of independent request to a model can significantly improve and reduce costs, as the processng can be done in parallel:

```python
response = model.batch([
    "Why do parrots are colorful",
    "How fish can swim?",
    "What is AI"
])

for res in response:
    print(res)
```

```python
response = model.batch([
    "Why do parrots are colorful",
    "How fish can swim?",
    "What is AI"
    ],
    config={
        'max_concurrency' : 5      # Limit to send at most 5 parallel calls at a time
    }
)

for res in response:
    print(res)
```
